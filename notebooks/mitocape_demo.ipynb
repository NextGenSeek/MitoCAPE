# MitoCAPE Demo

This notebook demonstrates how to use **MitoCAPE** for mitochondrial codon and amino-acid consequence analysis.

It covers:
- package import
- running a basic analysis
- inspecting outputs
- generating selected plots


## 1. Installation

If MitoCAPE is not yet installed, run one of the following commands in a terminal:

```bash
pip install git+https://github.com/NextGenSeek/MitoCAPE.git
```

or, from a local clone of the repository:

```bash
pip install .
```


## 2. Import MitoCAPE


````python
from pathlib import Path
from mitocape import run_analysis
````

## 3. Define input and output paths

Update `input_file` to match the location of your example file if needed.


````python
input_file = Path("../examples/i726-i507.csv")
outdir = Path("../results_demo")
species = "fly"

print("Input:", input_file)
print("Output directory:", outdir)
print("Species:", species)
````

## 4. Run a basic analysis

This example runs MitoCAPE using the default core plots:
- standard
- top positions
- bubble


````python
results = run_analysis(
    input_file=input_file,
    outdir=outdir,
    species=species,
    value_threshold=0,
    make_plots=True,
)

results
````

## 5. Inspect the processed output file


````python
import pandas as pd

processed_df = pd.read_csv(results["processed_file"])
processed_df.head()
````

## 6. View summary statistics


````python
results["summary"]
````

## 7. List generated plot files


````python
for plot_file in results["plots"]:
    print(plot_file)
````

## 8. Run selected plot types only

This example generates only a subset of plots.


````python
selected_results = run_analysis(
    input_file=input_file,
    outdir=outdir / "selected_plots",
    species=species,
    value_threshold=5,
    make_plots=True,
    plot_types=["standard", "bubble", "top_positions"],
)

selected_results["plots"]
````

## 9. Optional: generate a sequence logo

Use `sequence_logo_threshold` together with `plot_types=["sequence_logo"]`.


````python
logo_results = run_analysis(
    input_file=input_file,
    outdir=outdir / "sequence_logo",
    species=species,
    make_plots=True,
    plot_types=["sequence_logo"],
    sequence_logo_threshold=10,
)

logo_results["plots"]
````

## 10. Notes

- Replace the example input file with your own dataset as needed.
- Supported species currently include `human`, `mouse`, and `fly`.
- Custom FASTA and gene maps can be supplied through `fasta_path` and `gene_map_path`.
- Output plots are saved as PDF files in the chosen output directory.
